```python
# ==========================================================
# NOTEBOOK 05_query_test.py
# ==========================================================

%pip install google-generativeai faiss-cpu

import google.generativeai as genai
import pandas as pd
import numpy as np
import faiss

genai.configure(api_key="Your Gemini API Key")

index = faiss.read_index("/dbfs/FileStore/faiss/docs.index")
meta = pd.read_pickle("/dbfs/FileStore/faiss/metadata.pkl")

question = "What is annual leave?"

q = genai.embed_content(
    model="models/gemini-embedding-2",
    content=question,
    task_type="RETRIEVAL_QUERY"
)["embedding"]

q = np.array([q]).astype("float32")

D, I = index.search(q, 3)

context = ""

for idx in I[0]:
    context += meta.iloc[idx]["chunk_text"] + "\n\n"

prompt = f"""
Use only context.

Context:
{context}

Question:
{question}
"""

res = genai.GenerativeModel(
"gemini-2.0-flash"
).generate_content(prompt)

print(res.text)

```
