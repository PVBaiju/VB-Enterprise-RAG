```python

CATALOG = "vb_rag_dev"
SCHEMA = "vb_rag_schema"

#spark.sql(f"CREATE CATALOG IF NOT EXISTS {CATALOG}")
spark.sql(f"CREATE DATABASE IF NOT EXISTS {CATALOG}.{SCHEMA}")
spark.sql(f"CREATE VOLUME IF NOT EXISTS {CATALOG}.{SCHEMA}.docs")

spark.sql(f"""
CREATE TABLE IF NOT EXISTS {CATALOG}.{SCHEMA}.bronze_documents(
 document_id STRING,
 file_name STRING,
 content STRING,
 source_path STRING,
 ingested_at TIMESTAMP
)
""")

spark.sql(f"""
CREATE TABLE IF NOT EXISTS {CATALOG}.{SCHEMA}.silver_chunks(
 chunk_id STRING,
 document_id STRING,
 file_name STRING,
 chunk_text STRING,
 chunk_order INT,
 source_path STRING,
 ingested_at TIMESTAMP
)
""")

print("Setup completed")
```
