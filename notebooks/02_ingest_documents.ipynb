```python
# ==========================================================
# NOTEBOOK 02_ingest_documents.py
# Upload employee_handbook.txt into Databricks Volume:
# /Volumes/vb_rag_dev/vb_rag_schema/docs/employee_handbook.txt
# ==========================================================

import uuid
from datetime import datetime

FILE_PATH = "/Volumes/vb_rag_dev/vb_rag_schema/docs/employee_handbook.txt"

text = dbutils.fs.head(FILE_PATH, 1000000)

rows = [{
    "document_id": str(uuid.uuid4()),
    "file_name": "employee_handbook.txt",
    "content": text,
    "source_path": FILE_PATH,
    "ingested_at": datetime.now()
}]

spark.createDataFrame(rows).write.mode("append").saveAsTable(
"vb_rag_dev.vb_rag_schema.bronze_documents"
)

print("Document loaded")
```
