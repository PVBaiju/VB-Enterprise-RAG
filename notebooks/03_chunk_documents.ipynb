```python
# ==========================================================
# NOTEBOOK 03_chunk_documents.py
# ==========================================================

import uuid
from datetime import datetime

CHUNK_SIZE = 1200
OVERLAP = 200

pdf = spark.table(
"vb_rag_dev.vb_rag_schema.bronze_documents"
).toPandas()

rows = []

for _, row in pdf.iterrows():

    txt = row["content"]
    start = 0
    order = 1

    while start < len(txt):

        end = start + CHUNK_SIZE
        chunk = txt[start:end]

        rows.append({
            "chunk_id": str(uuid.uuid4()),
            "document_id": row["document_id"],
            "file_name": row["file_name"],
            "chunk_text": chunk,
            "chunk_order": order,
            "source_path": row["source_path"],
            "ingested_at": datetime.now()
        })

        start += CHUNK_SIZE - OVERLAP
        order += 1

spark.createDataFrame(rows).write\
    .mode("overwrite")\
    .option("overwriteSchema", "true")\
    .saveAsTable("vb_rag_dev.vb_rag_schema.silver_chunks"
)

print("Chunking completed")
```

```python
%sql
Select * from  vb_rag_schema.silver_chunks order by chunk_order
```
