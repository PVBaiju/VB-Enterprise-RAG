```python

# ==========================================================
# NOTEBOOK 04_generate_embeddings_and_index.py
# ==========================================================

%pip install google-generativeai faiss-cpu

import os
import google.generativeai as genai
import pandas as pd
import numpy as np
import faiss

GOOGLE_API_KEY = "Your Gemini API Key"

genai.configure(api_key=GOOGLE_API_KEY)

pdf = spark.table(
"vb_rag_dev.vb_rag_schema.silver_chunks"
).toPandas()

embeddings = []

for txt in pdf["chunk_text"]:

    r = genai.embed_content(
        model="models/gemini-embedding-2",
        content=txt,
        task_type="RETRIEVAL_DOCUMENT"
    )

    embeddings.append(r["embedding"])

pdf["embedding"] = embeddings

spark.createDataFrame(pdf).write.mode("overwrite").option(
"overwriteSchema","true"
).saveAsTable(
"vb_rag_dev.vb_rag_schema.silver_chunks"
)

X = np.array(embeddings).astype("float32")

index = faiss.IndexFlatL2(X.shape[1])
index.add(X)

# Create folder first
os.makedirs("/dbfs/FileStore/faiss", exist_ok=True)

# Save files
faiss.write_index(index, "/dbfs/FileStore/faiss/docs.index")
pdf.to_pickle("/dbfs/FileStore/faiss/metadata.pkl")

print("FAISS index created successfully")
```
