```python
# Databricks notebook source

CATALOG = "vb_rag_demo"
SCHEMA = "rag_demo"
VOLUME = f"/Volumes/{CATALOG}/{SCHEMA}/docs"

BRONZE_TABLE = f"{CATALOG}.{SCHEMA}.bronze_documents"
SILVER_TABLE = f"{CATALOG}.{SCHEMA}.silver_chunks"

INDEX_NAME = f"{CATALOG}.{SCHEMA}.docs_index"
VECTOR_SEARCH_ENDPOINT = "rag-demo-endpoint"
EMBEDDING_MODEL = "text-embedding-3-small"

# Create schema and volume (catalog already exists)
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.{SCHEMA}")
spark.sql(f"CREATE VOLUME IF NOT EXISTS {CATALOG}.{SCHEMA}.docs")

# Create Bronze Table
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {BRONZE_TABLE} (
  document_id STRING,
  file_name STRING,
  file_path STRING,
  file_type STRING,
  raw_text STRING,
  ingested_at TIMESTAMP
)
USING DELTA
""")

# Create Silver Table
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {SILVER_TABLE} (
  chunk_id STRING,
  document_id STRING,
  file_name STRING,
  chunk_text STRING,
  chunk_order INT,
  source_path STRING,
  ingested_at TIMESTAMP
)
USING DELTA
""")
```

```python
spark.sql("CREATE CATALOG vb_rag_demo")
spark.sql("CREATE SCHEMA vb_rag_demo.rag_demo")
spark.sql("CREATE VOLUME vb_rag_demo.rag_demo.docs")
```
